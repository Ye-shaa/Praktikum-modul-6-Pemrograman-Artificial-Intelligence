```python
#Contoh penggunaan Nested Loop

i = 2
while(i < 100):
    j = 2
    while(j <= (i/j)):
        if not(i%j): break
        j = j + 1
    if (j > i/j) : print(i, " is prime")
    i = i + 1

print("Good bye!")
```

```python
#Program 1
list_angka = [1, 2, 3, 4, 5, 6, 7, 8, 9]

jumlah_genap = 0
jumlah_ganjil = 0

for angka in list_angka:
    if angka % 2 == 0:
        jumlah_genap += 1
    else:
        jumlah_ganjil += 1

print("list angka,", tuple(list_angka))
print(f"Jumlah bilangan genap : {jumlah_genap}")
print(f"Jumlah bilangan ganjil : {jumlah_ganjil}")
```
